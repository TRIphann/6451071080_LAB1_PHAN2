```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_excel('Online Retail.xlsx')
df.columns = df.columns.str.strip()

# chuan hoa va loai bo hoa don tinh dung (ky tu 'C')
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C', na=False)]

# Loc du lieu nuoc Anh, gom nhom theo hoa don + ten mat hang
basket = (df[df['Country'] == "United Kingdom"]
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('InvoiceNo'))

print("Da tao xong bang gio hang (basket) thanh cong!")
print("Kich thuoc bang:", basket.shape)
print("So hoa don:", basket.shape[0], "| So mat hang:", basket.shape[1])
print("Tong so dong du lieu goc:", len(df))

print("\n--- Thong ke tong so luong mua theo mat hang (top 10) ---")
top_items = basket.sum().sort_values(ascending=False).head(10)
display(top_items)

print("\n--- Vi du: 5 hoa don dau tien (chi 8 mat hang dau) ---")
display(basket.iloc[:5, :8])
```
