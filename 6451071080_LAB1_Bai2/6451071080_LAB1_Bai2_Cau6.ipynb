```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_excel('Online Retail.xlsx')
df.columns = df.columns.str.strip()

# chuan hoa va loai bo hoa don tinh dung (ky tu 'C')
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C', na=False)]

# Gom nhom theo hoa don + ten mat hang (moi dong = 1 hoa don)
basket = (df[df['Country'] == "United Kingdom"]
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum()
          .unstack()
          .reset_index()
          .fillna(0)
          .set_index('InvoiceNo'))

print("Kich thuoc bang gio hang (bang so luong mua):", basket.shape)

# CHuyen doi sang HOT ENCODING: gia tri > 0 -> 1, con lai -> 0
hot_encoded = basket.map(lambda x: 1 if x > 0 else 0)

print("Kich thuoc bang HOT ENCODING:", hot_encoded.shape)
print("So hoa don:", hot_encoded.shape[0], "| So mat hang:", hot_encoded.shape[1])
print("Tong so gia tri 1 (so mat hang duoc mua):", int(hot_encoded.values.sum()))
print("Tong so gia tri 0:", int((hot_encoded.values == 0).sum()))

print("\n--- 5 hoa don dau tien sau khi HOT ENCODING (8 mat hang dau) ---")
display(hot_encoded.iloc[:5, :8])

print("\n--- Kiem tra: cot nao co so 1 lon nhat (mat hang ban chay nhat) ---")
display(hot_encoded.sum().sort_values(ascending=False).head(10))
```
