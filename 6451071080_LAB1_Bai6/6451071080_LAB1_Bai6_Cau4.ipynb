```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Doc du lieu: moi dong cua file la 1 giao dich, cac muc tach nhau bang dau phay
transactions = []
with open("groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [item.strip() for item in line.strip().split(",") if item.strip()]
        if items:
            transactions.append(items)

# Dem so lan xuat hien cua tung mat hang
flat_items = [item for t in transactions for item in t]
item_counts = pd.Series(flat_items).value_counts()

print("So mat hang duy nhat:", len(item_counts))
print("Tong so lan mat hang xuat hien:", int(item_counts.sum()))
print("\n--- 20 mat hang pho bien nhat ---")
display(item_counts.head(20))

# Bieu do thanh ngang: 20 mat hang pho bien nhat
plt.figure(figsize=(12, 8))
item_counts.head(20).sort_values().plot(kind="barh", color="skyblue")
plt.title("Top 20 mặt hàng phổ biến nhất trong dataset Groceries")
plt.xlabel("Tần suất xuất hiện")
plt.ylabel("Mặt hàng")
plt.tight_layout()
plt.show()
```
