```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder

# Doc du lieu: moi dong cua file la 1 giao dich, cac muc tach nhau bang dau phay
transactions = []
with open("groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [item.strip() for item in line.strip().split(",") if item.strip()]
        if items:
            transactions.append(items)

print("So giao dich:", len(transactions))

# CHuyen doi sang ONE-HOT ENCODING bang TransactionEncoder cua mlxtend
# ket qua: moi dong = 1 giao dich, moi cot = 1 mat hang, gia tri 0/1
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
basket = pd.DataFrame(te_ary, columns=te.columns_).astype(int)

print("Kich thuoc bang one-hot:", basket.shape)
print("So giao dich:", basket.shape[0], "| So mat hang:", basket.shape[1])
print("Tong so gia tri 1 (so lan mat hang xuat hien):", int(basket.values.sum()))
print("Tong so gia tri 0:", int((basket.values == 0).sum()))

print("\n--- 5 giao dich dau tien (10 mat hang dau) ---")
display(basket.iloc[:5, :10])

print("\n--- So mat hang trung binh moi giao dich ---")
print(round(basket.values.sum() / len(basket), 2))

print("\n--- 10 mat hang pho bien nhat (so giao dich co mua) ---")
display(basket.sum().sort_values(ascending=False).head(10))
```
