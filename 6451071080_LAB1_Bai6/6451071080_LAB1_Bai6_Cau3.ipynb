```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Doc du lieu: moi dong cua file la 1 giao dich, cac muc tach nhau bang dau phay
transactions = []
with open("groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [item.strip() for item in line.strip().split(",") if item.strip()]
        if items:
            transactions.append(items)

# Loai bo cac giao dich rong (khong co mat hang nao)
valid_transactions = [t for t in transactions if len(t) > 0]
print(f"Số lượng giao dịch hợp lệ sau khi xử lý: {len(valid_transactions)}")
print(f"Số giao dịch bị loại bỏ (rỗng): {len(transactions) - len(valid_transactions)}")

# Phan bo so mat hang trong moi giao dich
so_mat_hang = pd.Series([len(t) for t in valid_transactions])
print("\n--- Thống kê số mặt hàng trong mỗi giao dịch ---")
print(so_mat_hang.describe())

print("\n--- Phân bố số mặt hàng trong mỗi giao dịch ---")
display(so_mat_hang.value_counts().sort_index().head(15))

print("\n--- Giao dịch có nhiều mặt hàng nhất ---")
for t in sorted(valid_transactions, key=len, reverse=True)[:5]:
    print(f"({len(t)} mat hang): {t}")
```
